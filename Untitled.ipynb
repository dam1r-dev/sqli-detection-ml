```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix

# 1. Загрузка данных
train_df = pd.read_csv('../data/train_clean.csv')
train_aug_df = pd.read_csv('../data/train_augmented.csv')
test_obf_df = pd.read_csv('../data/test_obfuscated.csv')

# 2. Baseline модель (Обучение и Предсказание)
vec_base = TfidfVectorizer(analyzer='char', ngram_range=(2, 5), max_features=10000)
X_train_base = vec_base.fit_transform(train_df['query'].astype(str))
clf_base = LogisticRegression(max_iter=1000).fit(X_train_base, train_df['label'])

X_test_obf_base = vec_base.transform(test_obf_df['query'].astype(str))
y_test_obf = test_obf_df['label']
y_pred_obf = clf_base.predict(X_test_obf_base)

# 3. Augmented модель (Обучение и Предсказание)
vec_aug = TfidfVectorizer(analyzer='char', ngram_range=(2, 5), max_features=10000)
X_train_aug = vec_aug.fit_transform(train_aug_df['query'].astype(str))
clf_aug = LogisticRegression(max_iter=1000).fit(X_train_aug, train_aug_df['label'])

X_test_obf_aug = vec_aug.transform(test_obf_df['query'].astype(str))
y_pred_obf_aug = clf_aug.predict(X_test_obf_aug)

# 4. Визуализация и сохранение графиков
plt.style.use('seaborn-v0_8-whitegrid')
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Baseline Heatmap
cm_base = confusion_matrix(y_test_obf, y_pred_obf)
sns.heatmap(cm_base, annot=True, fmt='d', cmap='Reds', ax=axes[0], cbar=False,
            xticklabels=['Benign', 'SQLi'], yticklabels=['Benign', 'SQLi'])
axes[0].set_title('Baseline Model on Obfuscated Test\n(32% Evasion Rate)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Predicted Label')
axes[0].set_ylabel('True Label')

# Augmented Heatmap
cm_aug = confusion_matrix(y_test_obf, y_pred_obf_aug)
sns.heatmap(cm_aug, annot=True, fmt='d', cmap='Greens', ax=axes[1], cbar=False,
            xticklabels=['Benign', 'SQLi'], yticklabels=['Benign', 'SQLi'])
axes[1].set_title('Augmented Model on Obfuscated Test\n(1% Evasion Rate)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Predicted Label')
axes[1].set_ylabel('True Label')

plt.tight_layout()
plt.savefig('../results/confusion_matrices.png', dpi=300)
plt.show()
print("График успешно сохранен в results/confusion_matrices.png")
```
